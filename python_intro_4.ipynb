```python
# Item 1
item1_name = input("Enter the Item name: ")
item1_price = int(input("Enter the price {item1_name}:₹ "))
item1_qty = int(input("Enter the qty{item1_name}: "))

 #item total
item1_total=item1_price*item1_qty

# Item 2
item2_name = input("Enter the Item name: ")
item2_price = int(input("Enter the price {item2_name}:₹ "))
item2_qty = int(input("Enter the qty{item2_name}: "))

 #item total
item2_total=item2_price*item2_qty

# Item 3
item3_name = input("Enter the Item name: ")
item3_price = int(input(f"Enter the price {item3_name}:₹ "))
item3_qty = int(input("Enter the qty{item3_name}: "))

 #item total
item3_total=item3_price*item3_qty

#subtotal:
subtotal = item1_total+item2_total+item3_total

# calculate 10% discount:
discount_amount = subtotal*0.10

#Total_bill_amount 
total_bill_amount= subtotal - discount_amount

#calculate GST (12%):
gst_tax_amount = total_bill_amount*0.12

#get the grand total :
grand_total = total_bill_amount+gst_tax_amount
```

```python
print("=="*30)
print("            RECEIPT             ")
print("=="*30)
print(f"(item1_name) x (item1_qty): (item1_total:.2f)")
print(f"(item2_name) x (item2_qty): (item2_total:.2f)")
print(f"(item3_name) x (item3_qty): (item3_total:.2f)")
print("-"*30)
print(f"Subtotal Bill Amount: (subtotal:.2f)")
print(f"10% discount:-(discount_amount:.2f)")
print("_"*30)
print(f"Total Bill Amount: ${total_bill_amount: .2f}")
print(f"GST(12%): +${gst_tax_amount:.2f}")
print("_"*30)
print(f"Grand total : ${grand_total:.2f}")
```
