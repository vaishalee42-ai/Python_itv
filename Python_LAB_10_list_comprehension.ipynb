```python
trains = {
    "Rajdhani":120,
    "Duronto": 95,
    "Garib Rath":40,
    "Shatabdi": 60
}
```

```python
available_trains = {
    train: seats
    for train, seats in trains.items()
    if seats > 50
}
print("Train With More Than 50 Seats:")
print(available_trains)
```
