```python
name="Dhanush"
age=20
city="Chennai"
print (name, age, city)
```

```python
print ("The name of the person",name,"he is",age,"years old from city",city)
```

```python
#f-string method:
print(f"The name of the person is {name} and he is {age} years old from city {city}")
```

```python
#format method
print("The name of the person is {} and he is {} years old from city {}".format(name,age,city))
```

```python
#format method: Positional value
print("The name of the person is {1} and he is {0} years old from city {2}".format(age,name,city))
```

```python
#format method: keyword value
print("The name of the person is {name} and he is {age} years old from city {city}".format(name="Vaishu",age=21,city="Chennai"))
```
