```python
#comparision operator (Boolean output)
x=20
y=30
print(x>y) #greater than
print(x>=y) #greater than or equal to
print(x<y) #lesser than
print(x<=y) #lesser than or equal to
print(x==y) # x is equal to y
print(x!=y) #x is not equal to y

```

```python
#Logical operator: and, or, not
a=100
b=150
c=50
print(a>b and a>c) #returns true when both the conditions are true else return false
print(a>b or a>c)# returns true when at least one condition is true
print (not(a>b)) # return opposite of actual output
```
